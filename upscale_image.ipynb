```python
from repositories.upscale.google import GoogleUpscaleRepository
from PIL import Image
from dotenv import load_dotenv
load_dotenv(".env")
from settings import GOOGLE_APPLICATION_CREDENTIALS
_ = GOOGLE_APPLICATION_CREDENTIALS

upscale_repository = GoogleUpscaleRepository()
```
